```python
# Read the raw DNA file into a list of lines
with open("dna_raw.txt", "r") as file:
    lines = file.readlines()

# Each sequence takes two lines: ">id" followed by the sequence
dna = {}
for i in range(0, len(lines), 2):
    seq_id = lines[i][1:].strip()            # remove ">" and the newline
    sequence = lines[i + 1].strip().upper()  # uppercase so "a" and "A" count as the same
    dna[seq_id] = sequence

# Count how many times each letter appears in each sequence
dna_counts = {}
for seq_id, sequence in dna.items():
    counts = {}
    for letter in sequence:
        # get() returns the current count, or 0 if the letter hasn't been counted yet
        counts[letter] = counts.get(letter, 0) + 1
    dna_counts[seq_id] = counts

for seq_id, counts in dna_counts.items():
    print(f"{seq_id}: {counts}")
```

```python
import matplotlib.pyplot as plt

for seq_id, counts in dna_counts.items():
    letters = list(counts.keys())
    values = list(counts.values())

    plt.figure(figsize=(6, 4))
    plt.bar(letters, values)
    plt.title(f"Bokstavsfördelning för {seq_id}")
    plt.xlabel("Bokstav")
    plt.ylabel("Antal")
    plt.tight_layout()
    plt.show()

for seq_id, counts in dna_counts.items():
    print(f"{seq_id}: {counts}")
```
